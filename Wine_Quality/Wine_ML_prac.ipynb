```python
# Loading and Inspecting the Data

import pandas as pd 
wine_data=pd.read_csv('winequality-red.csv',sep=';')
print(wine_data.head(5))
print(wine_data.shape)
wine_data.info()
print(wine_data.describe())
print(wine_data.columns)
print("Mean:\n",wine_data.mean())
print("\nMedian:\n",wine_data.median())
print("\nStandard Deviation:\n",wine_data.std())



print(f" Number of Unique Lables in Target Column: {wine_data['quality'].nunique()}")
print(wine_data['quality'].unique())
print(wine_data.isnull().sum())
print(wine_data.duplicated().sum())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.histplot(data=wine_data,x='alcohol',bins=20,kde=True)
plt.show()
sns.histplot(data=wine_data,x='fixed acidity',bins=20,kde=True)
plt.show()
sns.histplot(data=wine_data,x='quality',bins=20,kde=True)
plt.show()
```

```python
num_col=wine_data.select_dtypes(include='number').columns
for col in num_col:
    q1=wine_data[col].quantile(0.25)
    q3=wine_data[col].quantile(0.75)
    iqr=q3-q1
    lower=q1-1.5 * iqr
    upper=q3+1.5 * iqr
    outliers=wine_data[(wine_data[col]<lower) | (wine_data[col]>upper)]
    print(f"Column: {col}")
    print(f"LOWER: {lower}")
    print(f"UPPER: {upper}")
    print(f"Total Number of Outliers: {len(outliers)}")
    print(outliers[col])
    wine_data[col].plot(kind='box',figsize=(8,5))
    plt.title(f'Box Plot - {col}')
    plt.show()
```

```python
# Featuring the Dataset
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
x=wine_data.drop(['quality'],axis=1)
y=wine_data['quality']
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
scaler=StandardScaler()
x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.transform(x_test)

print("Original Training Data:")
print(x_train.mean())

print("\nScaled Training Data:")
print(pd.DataFrame(x_train_scaled, columns=x.columns).mean())

print("\nScaled Standard Deviation:")
print(pd.DataFrame(x_train_scaled, columns=x.columns).std())
```

```python
# Prediction Using DecisionTreeRegressor Model
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
import numpy as np

dtr=DecisionTreeRegressor(random_state=42)
dtr.fit(x_train_scaled,y_train)
dtr_pred=dtr.predict(x_test_scaled)
print("MAE:", mean_absolute_error(y_test, dtr_pred))
mse=mean_squared_error(y_test,dtr_pred)
print("MSE:", mse)
print("RMSE:", np.sqrt(mse))
print("R2 Score:", r2_score(y_test, dtr_pred))
```

```python
# Predictions Using RandomForestRegressor Model
from sklearn.ensemble import RandomForestRegressor
rfr=RandomForestRegressor(n_estimators=100,random_state=42)
rfr.fit(x_train_scaled,y_train)
rfr_pred=rfr.predict(x_test_scaled)
print("MAE:", mean_absolute_error(y_test, rfr_pred))
mse=mean_squared_error(y_test,rfr_pred)
print("MSE:", mse)
print("RMSE:", np.sqrt(mse))
print("R2 Score:", r2_score(y_test, rfr_pred))
```

```python
# CrossValidation Model
from sklearn.model_selection import cross_val_score
cv_scores=cross_val_score(rfr,x_train_scaled,y_train,cv=5,scoring='r2')
print("Cross Validation Scores:", cv_scores)
print("Average R2:", cv_scores.mean())
```
